# Iris Dataset - Exploratory Data Analysis

This notebook provides a comprehensive exploratory data analysis (EDA) of the Iris dataset.

```python
# Setup
import sys
sys.path.insert(0, '../src')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from iris_classifier import IrisDataLoader, IrisVisualizer

%matplotlib inline
plt.style.use('seaborn-v0_8-darkgrid')
```

## 1. Load the Dataset

```python
# Load data
data_loader = IrisDataLoader()
X, y = data_loader.get_full_dataset()

print("Dataset loaded successfully!")
print(f"Shape: {X.shape}")
```

## 2. Dataset Information

```python
# Get dataset info
info = data_loader.get_dataset_info()

print("Dataset Information:")
print(f"  Number of samples: {info['n_samples']}")
print(f"  Number of features: {info['n_features']}")
print(f"  Number of classes: {info['n_classes']}")
print(f"\nFeatures: {info['feature_names']}")
print(f"\nClasses: {info['target_names']}")
print(f"\nClass Distribution: {info['class_distribution']}")
```

## 3. Feature Statistics

```python
# Display statistics
stats = data_loader.get_feature_statistics()
print("Feature Statistics:")
display(stats)
```

## 4. Visualizations

```python
# Initialize visualizer
visualizer = IrisVisualizer()
```

### 4.1 Class Distribution

```python
visualizer.plot_class_distribution(y)
```

### 4.2 Feature Distributions

```python
visualizer.plot_feature_distributions(X, y)
```

### 4.3 Correlation Matrix

```python
visualizer.plot_correlation_matrix(X)
```

### 4.4 Pairplot

```python
visualizer.plot_pairplot(X, y)
```

### 4.5 PCA Visualization

```python
visualizer.plot_pca_visualization(X, y)
```

## 5. Key Insights

Based on the exploratory data analysis:

1. **Balanced Dataset**: All three classes have equal representation (50 samples each)
2. **Clear Separability**: Setosa appears to be linearly separable from the other two species
3. **Petal Features**: Petal length and width show strong discriminative power
4. **Correlations**: High correlation between petal length and width suggests they carry similar information
5. **No Missing Values**: The dataset is complete with no missing values
